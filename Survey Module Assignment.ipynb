## Part A: Survey Design

**Survey Topic:** AI in education - What are students' attitudes toward AI tools in their coursework? How do they use these tools, and what concerns do they have?

### Research Objectives

The purpose of this survey is to understand undergraduate students’ attitudes toward artificial intelligence (AI) tools in their coursework. In recent years, generative AI tools like ChatGPT have become increasingly common in academic environments. Universities are still determining how these tools should be integrated into learning, and students are navigating how to use them while balancing concerns about academic integrity, learning outcomes, and fairness.

This survey aims to examine how frequently students use AI tools for coursework, what types of tasks they use them for, and how their usage relates to their attitudes about AI’s role in education. More specifically, the survey will explore whether students who use AI tools more frequently have more positive attitudes toward their use in academic work, and whether concerns about academic integrity or learning quality influence students’ willingness to use these tools. The results may help better understand how students perceive AI and how institutions might design policies or guidance around its use.

### Target Population & Sampling Strategy

The target population of this survey is current undergraduate students in the United States. Participants would be recruited through university and student organization mailing lists, posters put up around campus, and online student communities such as campus forums or social media groups (Facebook, GroupMe, Slack, etc.). Participation would be voluntary and anonymous, and the survey would be administered online through Google Forms or Qualtrics.

One practical challenge in obtaining a representative sample is that students who already have strong opinions about AI or who frequently use AI tools may be more likely to participate. In addition, recruiting through online channels may overrepresent students who are more technologically engaged. As a result, the sample may not perfectly reflect the broader undergraduate population.

### Design Rationale

To avoid leading language, double-barreled wording, and ambiguous terms each question was written to focus on single concept. For example, perceptions of AI’s benefits for learning and concerns about its impact on learning are asked in separate questions rather than combined into one. Neutral wording was used throughout to avoid implying that AI use is either positive or negative. The survey also refers broadly to “AI tools” and provides examples instead of assuming familiarity with specific platforms.

Response options were designed to be mutually exclusive and collectively exhaustive. Frequency questions use ordered categories that reflect realistic patterns of behavior, ranging from “never” to “weekly or more.” Attitudinal questions use a standard five-point Likert scale with a neutral midpoint, allowing respondents to express varying levels of agreement. A multi-select question was included to capture the different ways students may use AI tools for coursework.

The survey questions are ordered to create a logical flow and reduce potential priming effects. It begins with demographic questions for context, followed by questions about awareness and usage of AI tools, and then moves to attitudes and concerns. Open-ended questions appear at the end so that respondents’ earlier answers are not influenced by reflective prompts. Each question was designed to measure a specific construct, such as frequency of AI use, perceived usefulness, or concerns about academic integrity, to improve the validity of the survey.

### Potential Biases & Limitations

Several response biases may affect the results of this survey. Social desirability bias may lead some students to underreport their use of AI tools if they believe such behavior might be viewed negatively. Conversely, some respondents may overreport usage if they perceive AI use as technologically advanced or desirable. In addition, extreme response bias can distort the measurement of attitudes, as AI-use, especially in education, is a controversial topic.

The sampling strategy may also limit generalizability. Recruiting participants through online student communities may overrepresent students who are more comfortable with digital technologies, which could bias estimates of AI usage. The sample may also not be representative across universities, majors, or demographic groups.

With more resources, the study could improve representativeness by using stratified sampling across universities and academic disciplines. Additional measures, such as validated scales for technology attitudes or learning outcomes, could also strengthen measurement reliability.

**Research Question:** Identity and political attitudes: How do different social identities (race, gender, education, religion, region) relate to political preferences? Are there interaction effects between identities?

### Introduction

**Research Question:**

This analysis examines how social identities relate to political preferences using data from the 2020 American National Election Studies (ANES) Time Series Study. More specifically, it investigates how gender, education, race, religion, and region relate to party identification.

Political party identification is a key indicator of political preference in the United States. Social identities such as gender, education level, and race often shape individuals’ political attitudes and affiliations. Understanding these relationships is important because it helps explain the underlying factors that drive political polarization and party alignment in the U.S. By analyzing how social characteristics relate to party identification, we can also gain insight into the patterns and causes of political behavior, inform strategies for civic engagement and representation, and help  anticipate future shifts in the political landscape as demographics change. 

**Hypotheses:**

H1: Individuals with higher levels of education will tend to identify less strongly with the Republican Party.

H2: Women will identify more with the Democratic Party compared to men.

H3: Race will be associated with party identification.

H4: There will be an interaction effect between gender and education, where education influences party identification differently for men and women.

### Data Loading & Exploration

```python
# Access data file in Google Drive
# from google.colab import drive
# drive.mount('/content/drive')

# Import module, libraries, and packages
import os
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

# Print Drive contents to find data file
print(os.listdir('/content/drive/My Drive'))
```

```python
# Load the 2020 ANES Time Series data
df = pd.read_stata('/content/drive/My Drive/38034-0001-Data.dta', convert_categoricals=False)

# Rename variables
df = df.rename(columns={
    "V201231X": "party_id7",      # 7-point party ID scale
    "V201600": "sex",             # 1=male, 2=female
    "V201511X": "education",      # Education level
    "V201549X": "race",           # Race
    "V200010B": "wt_full_post"    # Full-sample post-election weight
})

# Female: from sex (1=male, 2=female)
df["female"] = (
    df["sex"]
      .replace({1: 0, 2: 1})
      .where(df["sex"].isin([1, 2]))
      .astype("Int64")
)

# Education should be numeric
df["education"] = pd.to_numeric(df["education"], errors="coerce")

# Race should be numeric
df["race"] = pd.to_numeric(df["race"], errors="coerce")

# Party ID: Keep only 1–7
df["party_id7"] = df["party_id7"].where(df["party_id7"].between(1, 7))
```

```python
# Create analysis dataframe
model_df = df[["party_id7","female","education","race","wt_full_post"]].dropna()

model_df.head()
```

### Descriptive Analysis

```python
# Summary statistics
model_df[["party_id7","female","education","race"]].describe()
```

```python
# Cross-tabulation
pd.crosstab(model_df["female"], model_df["party_id7"])
```

### Statistical Analysis

```python
model_pid = smf.wls(
    formula="party_id7 ~ female + education + race",
    data=model_df,
    weights=model_df["wt_full_post"],
).fit(cov_type='HC0')

print(model_pid.summary())
```

#### Visualization 1: 

```python
pid_labels = ["Strong Dem", "Weak Dem", "Lean Dem", "Independent", "Lean Rep", "Weak Rep", "Strong Rep"]

male_pct   = model_df[model_df.female==0].groupby("party_id7")["wt_full_post"].sum()
female_pct = model_df[model_df.female==1].groupby("party_id7")["wt_full_post"].sum()
male_pct   = male_pct   / male_pct.sum() * 100
female_pct = female_pct / female_pct.sum() * 100

x = np.arange(7)
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(x - 0.2, male_pct,   0.4, label="Male",   color="#4472C4")
ax.bar(x + 0.2, female_pct, 0.4, label="Female", color="#ED7D31")
ax.set_xticks(x)
ax.set_xticklabels(pid_labels, rotation=20, ha='right')
ax.set_ylabel("Weighted %")
ax.set_title("Figure 1: Party Identification by Gender")
ax.legend()
plt.tight_layout()
plt.show()
```
